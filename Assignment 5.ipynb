```python
import numpy as np

# Create a 1D array with 5 elements
arr = np.array([10, 20, 30, 40, 50])
print("Original Array:", arr)

# a) Addition of 2 in all elements
add_arr = arr + 2
print("a) After adding 2:", add_arr)

# b) Multiply 3 with all elements
mul_arr = arr * 3
print("b) After multiplying by 3:", mul_arr)

# c) Divide every element by 2
div_arr = arr / 2
print("c) After dividing by 2:", div_arr)
```

```python
import numpy as np

# a) Reverse the NumPy array
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print("a) Reversed array:", reversed_arr)


# b) Find the most frequent value and their indices
def find_most_frequent(arr_input):
    # bincount counts occurrences of non-negative integers
    counts = np.bincount(arr_input)
    most_frequent_val = np.argmax(counts)
    indices = np.where(arr_input == most_frequent_val)[0]
    return most_frequent_val, indices


# i. Array x
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
val_x, idx_x = find_most_frequent(x)
print(f"b) i. For array x: Most frequent value is {val_x} at indices {idx_x}")

# ii. Array y
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
# Array y has a tie (1 and 3 both appear 3 times). Let's handle multiple modes:
counts_y = np.bincount(y)
max_count = np.max(counts_y)
modes = np.where(counts_y == max_count)[0]

for mode in modes:
    idx_y = np.where(y == mode)[0]
    print(
        f"b) ii. For array y: Most frequent value is {mode} at indices {idx_y}"
    )
```

```python
import numpy as np

# 1. Create a 3x4 2D array
ucs420_your_name = np.array([[10, 20, 30, 40], [50, 60, 70, 80], [90, 15, 20, 35]])
print("Original 3x4 Array:\n", ucs420_your_name)

# 2. Compute Statistics
print("\n--- Statistics ---")
print("Mean:", np.mean(ucs420_your_name))
print("Median:", np.median(ucs420_your_name))
print("Max:", np.max(ucs420_your_name))
print("Min:", np.min(ucs420_your_name))
print("Unique elements:", np.unique(ucs420_your_name))

# 3. Reshape array to 4 rows and 3 columns
reshaped_ucs420_your_name = ucs420_your_name.reshape(4, 3)
print("\n--- Reshaped Array (4x3) ---\n", reshaped_ucs420_your_name)

# 4. Resize array to 2 rows and 3 columns
# np.resize returns a new copy, truncating data if it doesn't fit
resized_ucs420_your_name = np.resize(ucs420_your_name, (2, 3))
print("\n--- Resized Array (2x3) ---\n", resized_ucs420_your_name)
```
